```python
import numpy as np
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

#Dataset
x = np.array([-2, -1, 0, 1, 2]).reshape(-1, 1)
y = np.array([9, 2, 1, 6, 17])

#Polynomial features
poly = PolynomialFeatures(degree=2)
x_poly = poly.fit_transform(x)

#Train Model
model = LinearRegression()
model.fit(x_poly, y)

#Prediction
y_pred = model.predict(x_poly)

# Equation y = ax^2 + bx + c
c = model.intercept_
b = model.coef_[1]
a = model.coef_[2]

print(f"a: {a}")
print(f"b: {b}")
print(f"c: {c}")

# MSE
mse = mean_squared_error(y, y_pred)
print("Predicted values: ", y_pred)
print("Train MSE: ",mse)
```

```python
import matplotlib.pyplot as plt

# Generate x values for the fitted curve
x_fit = np.linspace(x.min(), x.max(), 100).reshape(-1, 1)
x_fit_poly = poly.transform(x_fit)
y_fit = model.predict(x_fit_poly)

# Plot original data points
plt.figure(figsize=(8, 6))
plt.scatter(x, y, color='blue', label='Original Data Points')

# Plot the fitted polynomial curve
plt.plot(x_fit, y_fit, color='red', label=f'Polynomial Regression (y = {a:.2f}x^2 + {b:.2f}x + {c:.2f})')

plt.title('Polynomial Regression of Degree 2')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.show()
```
