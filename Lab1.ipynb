```python
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Pandas Series
data = [30, 50, 40, 32]

series = pd.Series(data, index=['a', 'b', 'c', 'd'])
print(series)

# Math module
print(math.sqrt(25))
print(math.factorial(5))
print(math.pi)
print(math.e)
print(math.sin(math.radians(60)))

# NumPy arrays
m1 = np.array([[1, 2], [3, 4]])
print(m1)

m2 = np.array([[2, 4], [6, 7]])
print(m2)

print(m1 + m2)

print(np.zeros([3, 4]))

# Pandas DataFrame
df = pd.read_csv('lab1.csv')

print(df.head(10))
print(df.tail(2))
print(df.shape)

df.info()

print(df.describe())

print(df['gender'])
print(df[['gender', 'age']])

print(df.iloc[0])
print(df.loc[1])

print(df[df['age'] > 25])

# Line graph
x = [1, 2, 3, 4]
y = [40, 60, 70, 20]

plt.plot(
    x, y,
    marker='o',
    linestyle='-',
    color='r',
    label='line graph'
)

plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Basic Line Plot')
plt.legend()
plt.show()

# Bar graph
plt.bar(x, y, color='b')
plt.xlabel('Categories')
plt.ylabel('Values')
plt.show()

# Histogram
data = np.random.randn(1000)

plt.hist(
    data,
    bins=30,
    color='purple',
    edgecolor='black'
)

plt.xlabel('Values')
plt.ylabel('Frequency')
plt.show()

# Scatter plot
z = np.random.rand(50)
w = np.random.rand(50)

plt.scatter(z, w, color='red', marker='x')
plt.xlabel('1')
plt.ylabel('2')
plt.show()

# Pie chart
labels = ['Apple', 'Mango', 'Grapes', 'Cherry']
sizes = [30, 40, 50, 60]

plt.pie(
    sizes,
    labels=labels,
    autopct='%1.1f%%',
    colors=['blue', 'red', 'green', 'yellow']
)

plt.show()
```
