```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
#Load dataset
df = pd.read_csv("linear_regression_lab_15_records.csv")
```

```python
#Separate feature and target
x = df[["Hours_Studied"]]
y = df[["Marks"]]
```

```python
#Split into training data and testing data
x_train = x.iloc[:12]
x_test = x.iloc[12:]
y_train = y.iloc[:12]
y_test = y.iloc[12:]
```

```python
#Create and train model
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
#Regression equation
print("Slope: ",model.coef_[0])
print("Intercept: ",model.intercept_)
```

```python
#Predictions
y_train_pred = model.predict(x_train)
y_test_pred = model.predict(x_test)
```

```python
#MSE
train_mse = mean_squared_error(y_train, y_train_pred)
test_mse = mean_squared_error(y_test, y_test_pred)
print("Train MSE: ",train_mse)
print("Test MSE: ",test_mse)
```

```python
#Plot
plt.scatter(x_train, y_train, label = "Training Data",color = 'green')
plt.scatter(x_test, y_test,label = "Testing Data", color = 'pink')
plt.plot(x,model.predict(x),label = "Regression Line",color = 'red')

plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.title("Linear Regression")
plt.legend()
plt.show()
```
